# Public city segregation results

This notebook reviews the committed snapshot. It makes no downloads or Routes API calls. For a full offline recalculation, run `python -m scripts.reproduce_public --maps --rebuild-neighbors` from the repository root.


```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'reproducibility/snapshot_manifest.json').exists())
BASE = ROOT / 'docs/outputs/multicity/walk_access_revision'
scores = pd.read_csv(BASE / 'city_scores.csv')
display(scores)
```

The normalization is `(score - K)/(1 - K)`. NYC uses 70 sections and subway/SIR only; the other cities use individual populated tract portions and their documented rail/bus feeds. The four categories overlap.
